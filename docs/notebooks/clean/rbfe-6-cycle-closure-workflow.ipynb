# RBFE on a ligand set with cycle closure

End-to-end workflow for a congeneric series:

1. Load the BRD protein and three ligands, register them on the data platform
2. Submit `RBFE(protein=..., ligands=..., fep_abfe=...)` — Konnektor plans pairs, then system-prep, FEP per edge, and cycle closure
3. Quote, confirm, and watch progress
4. Inspect pairwise ΔΔG results and per-ligand absolute dG from cycle closure


## Setup

```python
from deeporigin.drug_discovery import (
    BRD_DATA_DIR,
    Ligand,
    LigandSet,
    Protein,
    RBFE,
    RBFEParams,
)
from deeporigin.platform import DeepOriginClient

client = DeepOriginClient()
client
```

## 1. Load structures and register on the data platform

We use the BRD4 example protein and three congeneric ligands from the bundled
dataset. `sync()` uploads files (when needed) and registers records on the
data platform. Konnektor mode requires every ligand to have a platform `id` and
remote `file_path`.

```python
protein = Protein.from_file(BRD_DATA_DIR / "brd.pdb")
protein.sync(client=client)
protein.id
```

```python
ligands = []
for ligand_file in ["brd-2.sdf", "brd-3.sdf", "brd-4.sdf"]:
    ligand = Ligand.from_sdf(BRD_DATA_DIR / ligand_file)
    ligand.sync(client=client)
    ligands.append(ligand)

ligand_set = LigandSet(ligands=ligands)
ligand_set
```

## 2. Submit RBFE workflow (Konnektor → system-prep → FEP → cycle closure)

Pass the ligand set (or a plain list). When `fep_abfe` and/or `exp_abfe`
anchors are provided, the SDK appends the `cycle-closure` step automatically.
Use at least one anchor ligand with a placeholder or measured absolute binding
free energy (dG in kcal/mol).

`test_run=1` shortens the simulation for exploration; use `test_run=0` for
production-quality results.


### Cycle-closure anchor

Anchor the network with a placeholder FEP-ABFE value for the first ligand.
Replace `-10.0` with a measured or computed ABFE when available.


```python
anchor_ligand = ligands[0]
fep_abfe = [{"ligand_id": anchor_ligand.id, "dG": -10.0}]
fep_abfe
```

## 3. Quote, confirm, and watch

```python
rbfe = RBFE(
    protein=protein,
    ligands=ligand_set,
    network_type="mst",
    params=RBFEParams(test_run=1),
    fep_abfe=fep_abfe,
    client=client,
)
assert rbfe.steps == ["konnektor", "system-prep", "rbfe", "cycle-closure"]
rbfe
```

```python
rbfe.start(quote=True)
rbfe.estimate
```
