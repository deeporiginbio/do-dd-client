# ADME properties 

## Load a ligand set

Here we load the BRD data set. Note that they have a custom column (`r_exp_dg`)

```python
ligands = LigandSet.from_dir(BRD_DATA_DIR)
ligands.sync()
```

```python
ligands.to_dataframe()
```

## Estimate and run molprops

In this section we create a `Molprops` object and use it to estimate and run these ligands:

```python
mp = Molprops(ligands=ligands)
mp.properties
```

```python
mp.run(quote=True)
```

```python
mp.run()
```

```python
mp.status
```

```python
ligands.to_dataframe()
```
