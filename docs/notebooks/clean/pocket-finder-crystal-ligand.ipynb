# PocketFinder from-crystal-ligand

Build one pocket from a crystal ligand using `PocketFinder` with `mode="from-crystal-ligand"`. Two ways to point at the ligand:

- `ligand_id`: a bare residue/ligand code resolved directly within a holo protein file (ligand still embedded).
- `crystal_ligand`: a `Ligand` already extracted from the protein (e.g. via `Protein.extract_ligand()`).

## Setup

```python
from deeporigin.drug_discovery import PocketFinder, Protein
```

## Load a holo protein

PDB `1eby` is a holo structure with its crystal ligand (residue code `BEB`) still embedded. We remove water only, so the ligand stays in place.

```python
protein = Protein.from_pdb_id("1eby")
protein.remove_water()
protein.sync()
protein.id
```

## Option 1: resolve the ligand in-place

Pass the bare ligand/residue code via `ligand_id`; the platform locates it directly within the supplied protein file, so no separate ligand file is needed.

```python
pf = PocketFinder(
    protein,
    mode="from-crystal-ligand",
    ligand_id="BEB",
)
pf
```

## Run

`run()` blocks until the tool finishes and returns a list of `Pocket` objects (one, PCA-aligned to the ligand's heavy-atom extents).

```python
pockets = pf.run()
pockets
```

```python
protein.show(pockets=pockets)
```

```python
pocket = pockets[0]
```

```python
pocket.show()
```

```python
pocket.show_box()
```

## Option 2: extract the ligand first

`Protein.extract_ligand()` pulls the crystal ligand out as a `Ligand` and strips it from the protein structure. Pass that `Ligand` in via `crystal_ligand` instead of `ligand_id`.

```python
crystal_ligand = protein.extract_ligand()
crystal_ligand.sync()
pf2 = PocketFinder(
    protein,
    mode="from-crystal-ligand",
    crystal_ligand=crystal_ligand,
)
pockets2 = pf2.run()
pockets2
```

## Visualize

```python
protein.show(pockets=pockets)
```
