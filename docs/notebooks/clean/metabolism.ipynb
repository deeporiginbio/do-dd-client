# Metabolism

The `Metabolism` class scores ligands for sites of metabolism on nine cytochrome P450 (CYP) isoforms. `run()` blocks and returns a table of sites. `get_molecules()` returns one reliability tier (`high`, `medium`, or `low`) per scored ligand.

There is no cost quote and no `start()`. SMILES are scored as written (atom indices are 0-based on that string).

```python
from deeporigin.drug_discovery import Metabolism, Ligand
from deeporigin.platform import DeepOriginClient

client = DeepOriginClient()
client
```

## Score a ligand

Construct `Metabolism` with a ligand (or a list, or a `LigandSet`). After construct, `enzymes` lists the nine CYP isoforms. Trim that list before `run()` if you only want some isoforms in the sites table.

```python
ligand = Ligand.from_smiles("CCO")
job = Metabolism(ligands=ligand, client=client)
job.enzymes
```

```python
job.enzymes = ["CYP3A4", "CYP2D6"]
sites = job.run()
sites
```

## Molecule-level confidence

`get_molecules()` is not filtered when you trim `enzymes`.

```python
mols = job.get_molecules()
mols
```
