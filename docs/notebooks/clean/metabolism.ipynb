# Metabolism

The `Metabolism` class scores ligands for sites of metabolism on cytochrome P450 (CYP) isoforms.

- `run()` blocks for small batches (fewer than 30 ligands) and returns a table of sites for every enzyme the tool scored.
- For 30 or more ligands, call `start()`, then `wait()` or `await watch()`, then `get_results()`.
- `get_molecules()` returns one reliability tier (`high`, `medium`, or `low`) per scored ligand.

There is no cost quote. SMILES are scored as written (atom indices are 0-based on that string).

```python
from deeporigin.drug_discovery import Metabolism, Ligand
from deeporigin.platform import DeepOriginClient

client = DeepOriginClient()
client
```

## Score a ligand

Construct `Metabolism` with a ligand (or a list, or a `LigandSet`) and call `run()`. The sites table includes every enzyme the tool scored.

```python
ligand = Ligand.from_smiles("CCO")
job = Metabolism(ligands=ligand, client=client)
sites = job.run()
sites
```

## Molecule-level confidence

`get_molecules()` returns one reliability tier per scored ligand.

```python
mols = job.get_molecules()
mols
```
