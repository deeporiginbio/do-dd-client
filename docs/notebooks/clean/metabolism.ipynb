# Site of Metabolism Tool

The `Metabolism` class scores ligands for sites of metabolism on cytochrome P450 (CYP) isoforms:

- CYP1A2
- CYP2A6
- CYP2B6
- CYP2C8
- CYP2C9
- CYP2C19
- CYP2D6
- CYP2E1
- CYP3A4


```python
from deeporigin.drug_discovery import Metabolism, Ligand, LigandSet, DATA_DIR
from deeporigin.platform import DeepOriginClient

client = DeepOriginClient()
client
```

## Score small ligandsets

In this section, we score a small ligand set. Small ligandsets can be scored in a synchronous (blocking) operation. Here, we pick the first 10 ligands:

```python
ligands = LigandSet.from_csv(DATA_DIR / "ligands" / "1000-ligands.csv")
ligands = ligands[:10]
ligands
```

```python
ligands.sync()
job = Metabolism(ligands=ligands, client=client)
```

```python
job.run()
```

```python
df = Metabolism.fetch_results(ligands)
df
```

## Molecule-level confidence

`get_molecules()` returns one reliability tier per scored ligand.

```python
df = Metabolism.fetch_molecules(ligands)
df
```

## Score large ligandsets (async)

For 30 or more ligands, use `start()` then `wait()` (or `await job.watch()` in a notebook). This runs on a provisioned instance, and can scale to large ligandsets. 

```python
ligands = LigandSet.from_csv(DATA_DIR / "ligands" / "1000-ligands.csv")
ligands = ligands[:100]
ligands.sync()
```

```python

job = Metabolism(ligands=ligands, client=client)
```

```python
job.start()
```

## Get results for a ligandset

To fetch results for a ligandset, across all exeuctions, we can use:

```python
df = Metabolism.fetch_results(ligands)
df
```

```python
df = Metabolism.fetch_molecules(ligands)
df
```
