# RBFE (just using Konnektor)

In this notebook we show how to use Konnektor as a stand-alone tool

## Setup

```python
from deeporigin.drug_discovery import (
    BRD_DATA_DIR,
    Pose,
    Protein,
    RBFEParams,
    Konnektor,
)
from deeporigin.platform import DeepOriginClient

client = DeepOriginClient()
client
```

```python
from deeporigin import projects
projects.create("python-client-dev")
```

## 1. Load structures and register on the data platform

We use the BRD4 example protein and three congeneric ligand poses from the
bundled dataset. `sync()` and `Pose.from_sdf()` upload files (when needed)
and register records on the data platform. Konnektor takes one pose per ligand.

```python
protein = Protein.from_file(BRD_DATA_DIR / "brd.pdb")
protein.sync()
protein.id
```

```python
poses = [
    Pose.from_sdf(BRD_DATA_DIR / ligand_file, protein_id=protein.id)
    for ligand_file in ["brd-2.sdf", "brd-3.sdf", "brd-4.sdf"]
]

poses
```

## Generate a network using Konnektor



```python
kn = Konnektor(poses=poses)
network = kn.run()
```

```python
network.show_network()
```
