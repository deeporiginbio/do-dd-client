# Structure Report

Grade a protein structure with `StructureReport`. Provide a local `Protein`, a four-character PDB ID, or both. `run()` blocks until the platform returns graded rows.


```python
from deeporigin.drug_discovery import Protein, StructureReport
from deeporigin.platform import DeepOriginClient

client = DeepOriginClient.from_disk()
client
```

## Remote PDB ID only

No coordinate upload — the tool scores from RCSB metadata.


```python
remote = StructureReport(pdb_id="1CRN", client=client)
remote_rows = remote.run()
remote_rows[0]
```

```python
row = remote_rows[0]
row.grade, row.weighted_score, row.metadata_source, row.pdb_id
```

## Local protein (+ optional PDB ID)

Sync a local structure, then grade it. Passing `pdb_id` uses RCSB as authority for experimental metadata.


```python
protein = Protein.from_pdb_id("1CRN")
protein.sync(client=client)
local = StructureReport(protein=protein, pdb_id="1CRN", client=client)
local_rows = local.run()
local_rows[0].grade, local_rows[0].metadata_source, local_rows[0].source_sha256 is not None
```
