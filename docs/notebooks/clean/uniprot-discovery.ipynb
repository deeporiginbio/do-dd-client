# UniProt discovery

Map a UniProtKB accession to ranked experimental PDB candidates with `UniprotDiscovery`, then import the recommended (or selected) structures into a project.


```python
from deeporigin.drug_discovery import Protein, UniprotDiscovery
from deeporigin.platform import DeepOriginClient

client = DeepOriginClient.from_disk()
client
```

## Browse candidates

`run()` returns ranked rows; exactly one is `recommended` unless the list is empty.


```python
job = UniprotDiscovery(uniprot_accession="P00533", client=client)
candidates = job.run()
candidates[:3]
```

```python
recommended = next(c for c in candidates if c.recommended)
recommended.pdb_id, recommended.grade, recommended.weighted_score
```

## Import recommended into a project

Requires a resolvable `project_id` (argument or `client.project_id`).


```python
proteins = job.import_proteins(project_id=client.project_id)
proteins[0].pdb_id, proteins[0].uniprot_accession, proteins[0].id
```

## Protein.from_uniprot sugar

One-liner for the recommended path.


```python
protein = Protein.from_uniprot(
    "P00533",
    project_id=client.project_id,
    client=client,
)
protein.pdb_id, protein.uniprot_accession
```
