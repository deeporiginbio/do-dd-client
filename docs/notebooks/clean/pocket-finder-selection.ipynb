# PocketFinder define-by-selection

Build one pocket from a residue (or ligand/cofactor) selection plus radius,
optionally PCA-aligned, using `PocketFinder` with `mode="define-by-selection"`.

## Setup

```python
from deeporigin.drug_discovery import BRD_DATA_DIR, PocketFinder, Protein
```

## Load protein

Use the example BRD protein. Remove water so author residue numbers match the
structure you sync to the platform.

```python
protein = Protein.from_file(BRD_DATA_DIR / "brd.pdb")
protein.remove_water()
protein.sync()
protein.id
```

## Define pocket by residue selection

Each selection needs `kind` and `author.chain_id`. Add `resseq` / `resname` so
the platform can exact-match one residue instance.

```python
selections = [
    {
        "kind": "residue",
        "author": {"chain_id": "A", "resseq": 55, "resname": "LYS"},
    }
]

pf = PocketFinder(
    protein,
    mode="define-by-selection",
    selections=selections,
    pocket_radius=10.0,
    align_to_pocket=True,
)
pf
```

## Run

`run()` blocks until the tool finishes and returns a list of `Pocket` objects
(typically one). Classifier score fields may be null for selection pockets.

```python
pockets = pf.run()
pockets
```

```python
pocket = pockets[0]
pocket.center, pocket.box
```

## Visualize

```python
protein.show(pockets=pockets)
```
